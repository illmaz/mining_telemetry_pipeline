```python
df = spark.readStream.format("cloudFiles")\
    .option("cloudFiles.format", "json")\
    .option("cloudFiles.schemaLocation","/Volumes/mining/bronze/checkpoints/schema")\
    .option("cloudFiles.schemaEvolutionMode", "addNewColumns")\
    .load("/Volumes/mining/landing/telemetry/")
```

```python
df.writeStream\
    .option("checkpointLocation","/Volumes/mining/bronze/checkpoints/checkpoint")\
    .option("mergeSchema", "true")\
    .trigger(availableNow=True)\
    .toTable("mining.bronze.haul_events")
```

```python
spark.read.table("mining.bronze.haul_events").groupBy("shift_id").count().show()
```
