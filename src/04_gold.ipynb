```python
from pyspark.sql.functions import sum, col, round as spark_round
```

```python
df = spark.read.table("mining.silver.haul_events").filter(col("shift_id").isNotNull())
```

```python
gold_df = df.groupBy("shift_id", "truck_id").agg(spark_round(sum("tonnage_num"), 1).alias("total_tonnage"))
```

```python
gold_df.show()
```

```python
gold_df.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable("mining.gold.production_reconciliation")
```
